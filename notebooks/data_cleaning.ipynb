```python
import pandas as pd
import numpy as np
```

```python
df_raw = pd.read_csv('D:/project_mindx/data/raw/superstore.csv')
print(f'số dòng và cột của dataset ban đầu: {df_raw.shape}')
```

```python
df_raw.info()
df_raw.columns
```

```python
#đổi tên cột cho dễ sử dụng
df_raw.rename(columns={'Order.Date': 'Order Date', 'shipping_cost': 'Shipping Cost', 'Ship.Mode': 'Ship Mode', 'Sub.Category': 'Sub Category', 'Product.Name': 'Product Name'}, inplace=True)
```

```python
#tạo data frame mới chứa cột chọn lọc
df = df_raw[['Order Date', 'Ship Mode', 'Segment', 'Country', 'Market', 'Category', 'Sub Category', 'Product Name', 'Quantity', 'Sales', 'Discount', 'Profit', 'Shipping Cost', 'Year']].copy()
```

```python
# Kiểm tra dòng trùng lặp hoàn toàn
duplicate_count = df.duplicated().sum()
print(f"Số dòng trùng lặp: {duplicate_count}")
if duplicate_count > 0:
    df.drop_duplicates(inplace=True)
    print("-> Đã xóa các dòng trùng lặp!")
# Kiểm tra giá trị thiếu (Null / NaN) ở từng cột
print("\nSố lượng giá trị Null trên từng cột:")
print(df.isnull().sum())
# Xóa các dòng có giá trị null (nếu có)
df.dropna(inplace=True)
```

```python
# Chuyển đổi cột Order Date sang kiểu datetime
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
```

```python
# Chuẩn hóa các cột số (ép kiểu float, làm tròn 2 chữ số thập phân)
numeric_cols = ['Sales', 'Profit', 'Shipping Cost']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce').round(2)
```

```python
# Thống kê tổng quan các cột số
df.describe()
```

```python
# Kiểm tra số lượng phân loại ở các cột chính
print("Danh mục Category:", df['Category'].unique())
print("Phân khúc Segment:", df['Segment'].unique())
print("Số quốc gia:", df['Country'].nunique())
print("Các năm có trong dữ liệu:", sorted(df_raw['Year'].dropna().unique()))
```

```python
df.to_csv('D:/project_mindx/data/processed/clean_superstore.csv', index=False, encoding='utf-8')
```
